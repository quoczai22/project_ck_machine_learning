```python
import pandas as pd

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv(
    "../data/raw/pima-indians-diabetes.csv",
    header=None,
    names=columns
)

df.head()
```

```python
print("=" * 60)
print("1. KIỂM TRA KÍCH THƯỚC DATASET")
print("=" * 60)

print(f"Số dòng dữ liệu : {df.shape[0]}")
print(f"Số cột dữ liệu  : {df.shape[1]}")
print(f"Số Feature      : {df.shape[1] - 1}")
print(f"Target          : Outcome")

if df.shape == (768, 9):
    print("\nKết luận: Dataset đúng kích thước yêu cầu 768 dòng × 9 cột.")
else:
    print("\nKết luận: Dataset không đúng kích thước mong đợi.")
```

```python
print("=" * 60)
print("2. KIỂM TRA KIỂU DỮ LIỆU")
print("=" * 60)

dtype_table = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Kiểu dữ liệu": df.dtypes.astype(str).values
})

dtype_table["Ý nghĩa kiểu"] = dtype_table["Kiểu dữ liệu"].replace({
    "int64": "Số nguyên",
    "float64": "Số thực"
})

display(dtype_table)

print("Kết luận: Tất cả các thuộc tính đều ở dạng số,")
print("phù hợp cho các bước phân tích và xây dựng mô hình học máy.")
```

```python
print("=" * 60)
print("3. KIỂM TRA GIÁ TRỊ THIẾU - NaN")
print("=" * 60)

missing_table = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Số giá trị thiếu": df.isnull().sum().values
})

missing_table["Tỷ lệ thiếu (%)"] = (
    df.isnull().sum().values / len(df) * 100
).round(2)

missing_table["Trạng thái"] = missing_table["Số giá trị thiếu"].apply(
    lambda x: "Không thiếu" if x == 0 else "Có dữ liệu thiếu"
)

display(missing_table)

total_missing = df.isnull().sum().sum()

print(f"Tổng số giá trị NaN: {total_missing}")

if total_missing == 0:
    print("Kết luận: Không phát hiện giá trị NaN trong dataset.")
else:
    print("Kết luận: Dataset có giá trị thiếu và cần được xem xét.")
```

```python
print("=" * 60)
print("4. KIỂM TRA DÒNG TRÙNG LẶP")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print(f"Tổng số dòng        : {len(df)}")
print(f"Số dòng trùng lặp   : {duplicate_count}")
print(f"Số dòng không trùng : {len(df) - duplicate_count}")

if duplicate_count == 0:
    print("\nKết luận: Không phát hiện dòng dữ liệu trùng lặp.")
else:
    print("\nKết luận: Dataset có dòng bị trùng lặp.")
    display(df[df.duplicated()])
```

```python
print("=" * 60)
print("5. KIỂM TRA BIẾN MỤC TIÊU OUTCOME")
print("=" * 60)

class_count = df["Outcome"].value_counts().sort_index()
class_percent = (
    df["Outcome"].value_counts(normalize=True).sort_index() * 100
).round(2)

outcome_table = pd.DataFrame({
    "Outcome": class_count.index,
    "Ý nghĩa": [
        "Không mắc tiểu đường",
        "Mắc tiểu đường"
    ],
    "Số lượng": class_count.values,
    "Tỷ lệ (%)": class_percent.values
})

display(outcome_table)

print(f"Tổng số mẫu: {class_count.sum()}")

valid_outcome = set(df["Outcome"].unique()).issubset({0, 1})

print(f"Outcome chỉ gồm 0 và 1: {valid_outcome}")

print("\nKết luận:")
print("- Outcome = 0 có 500 mẫu, chiếm khoảng 65.10%.")
print("- Outcome = 1 có 268 mẫu, chiếm khoảng 34.90%.")
print("- Có sự chênh lệch về số lượng giữa hai lớp.")
```

```python
print("=" * 60)
print("6. KIỂM TRA GIÁ TRỊ BẰNG 0")
print("=" * 60)

zero_count = (df == 0).sum()
zero_percent = ((df == 0).sum() / len(df) * 100).round(2)

zero_table = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Số giá trị 0": zero_count.values,
    "Tỷ lệ 0 (%)": zero_percent.values
})

display(zero_table)
```

```python
print("NHẬN XÉT GIÁ TRỊ 0")
print()

print("Pregnancies = 0:")
print("→ Có thể xảy ra, không xem là dữ liệu thiếu.")

print("\nOutcome = 0:")
print("→ Là nhãn hợp lệ của bài toán, không phải dữ liệu thiếu.")

print("\nCác thuộc tính cần chú ý:")
print("- Glucose: 5 giá trị bằng 0")
print("- BloodPressure: 35 giá trị bằng 0")
print("- SkinThickness: 227 giá trị bằng 0")
print("- Insulin: 374 giá trị bằng 0")
print("- BMI: 11 giá trị bằng 0")

print("\nKết luận:")
print("Các giá trị 0 ở những thuộc tính trên cần được xem xét")
print("ở bước tiền xử lý dữ liệu.")
print("Task 2 chỉ phát hiện và thống kê, chưa thay đổi dữ liệu raw.")
```

```python
print("=" * 60)
print("7. THỐNG KÊ MÔ TẢ DỮ LIỆU")
print("=" * 60)

statistics = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Min": df.min().values,
    "Max": df.max().values,
    "Trung bình": df.mean().round(2).values,
    "Trung vị": df.median().round(2).values,
    "Độ lệch chuẩn": df.std().round(2).values
})

display(statistics)
```
