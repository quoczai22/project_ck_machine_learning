```python
import pandas as pd
import hashlib

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv(
    "../data/raw/pima-indians-diabetes.csv",
    header=None,
    names=columns
)

df.head()
```

```python
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])
```

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Số feature:", X.shape[1])
print("Target:", y.name)
```

```python
profile = pd.DataFrame({
    "ThongTin": [
        "Tên dataset",
        "Số dòng",
        "Số cột",
        "Số feature",
        "Target",
        "Số mẫu Outcome = 0",
        "Số mẫu Outcome = 1"
    ],

    "KetQua": [
        "Pima Indians Diabetes",
        df.shape[0],
        df.shape[1],
        df.shape[1] - 1,
        "Outcome",
        (df["Outcome"] == 0).sum(),
        (df["Outcome"] == 1).sum()
    ]
})

profile
```

```python
profile.to_csv(
    "../reports/data_profile.csv",
    index=False,
    encoding="utf-8-sig"
)
```
